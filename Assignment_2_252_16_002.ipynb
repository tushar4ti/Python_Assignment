15. Write a program that takes a student&#39;s marks and first checks if the marks are above or
below 40 (pass/fail). If passed, use a nested if to decide the grade (A, B, or C).

```python
def result(mark):
  if (mark >= 0 and mark <= 100):
    if (mark >= 90 and mark <= 100):
      grade="A+"
    elif (mark >= 80 and mark <= 89):
      grade="A"
    elif (mark >= 60 and mark <= 79):
      grade="B"
    elif (mark >= 40 and mark <= 59):
      grade="C"
    else:
      grade="F"

    if (grade == "F"):
      print("Fail")
    else:
      print("Pass")
      print(f"Grade: {grade}")
  else:
    print("Invalid marks")
mark=int(input("Enter your marks : "))
result(mark)
```
